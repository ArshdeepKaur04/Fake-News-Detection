## **Importing Necessary Libraries**

```python
import os
import numpy as np
import pandas as pd
from tqdm import tqdm
from sentence_transformers import SentenceTransformer
```

## **Loading the Preprocessed Data**

```python
df = pd.read_csv(r"E:\Arshdeep CL NLP Project\Dataset\preprocessed\fake_news_cleaned.csv")
```

```python
df.head(10)
```

## **Load the Embedding Model + Caching (For Faster Processing)**

```python
model = SentenceTransformer('all-MiniLM-L6-v2')
```

## **Generating Embeddings, Arshdeep!**

```python
df = df.dropna(subset=["lemmatized_text", "label"]) # No NaN values should be there
```

### Extracting Inputs and Targets

```python
texts = df["lemmatized_text"].tolist()
labels = df["label"].tolist()
```

```python
embeddings = model.encode(texts, batch_size=64, show_progress_bar=True)

embeddings = np.array(embeddings)

print(f"Embeddings shape: {embeddings.shape}")
```

## **Saving for Training Further, Arshdeep!**

```python
os.makedirs(r"E:\Arshdeep CL NLP Project\Dataset/preprocessed", exist_ok=True)

np.save(r"E:\Arshdeep CL NLP Project\Dataset/preprocessed/embeddings.npy", embeddings)
np.save(r"E:\Arshdeep CL NLP Project\Dataset/preprocessed/labels.npy", np.array(labels))

print("✅ Saved embeddings and labels, Arshdeep!")
```
